# 1. Importing Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2. Loading and Previewing the Raw Dataset

```python
df = pd.read_csv("../Dataset/Raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")

df.head()
```

# 3. Identifying Blank Values in TotalCharges

```python
df["TotalCharges"].str.strip().eq("").sum()
```

```python
df[df["TotalCharges"].str.strip() == ""]
```

# 4. Converting TotalCharges to Numeric Data Type

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
df["TotalCharges"].isnull().sum()
```

# 5. Handling Missing Values in TotalCharges

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
df["TotalCharges"].isnull().sum()
```

# 6. Verifying TotalCharges Data Type

```python
df["TotalCharges"].dtype
```

```python
df.isnull().sum()
```

# 7. Checking for Duplicate Records

```python
df.duplicated().sum()
```

# 8. Reviewing Updated Data Types

```python
df.dtypes
```

# 9. Inspecting Unique Values in Categorical Columns

```python
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

# 10. Validating the Cleaned Dataset

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Duplicate Rows:", df.duplicated().sum())
print("Missing Values:", df.isnull().sum().sum())
```

# 11. Verifying the Processed Dataset File

```python
# Export the cleaned dataset
df.to_csv("../Dataset/Processed/Customer_Churn_Cleaned.csv", index=False)

print("Cleaned dataset exported successfully.")
```
