# 1. Importing Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2. Loading and Previewing the Cleaned Dataset

```python
df = pd.read_csv("../Dataset/Processed/Customer_Churn_Cleaned.csv")

df.head()
```

# 3. Checking Dataset Dimensions and Data Types

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nData Types:")
print(df.dtypes)
```

```python
df[["SeniorCitizen", "tenure", "MonthlyCharges", "TotalCharges"]].describe()
```

# 4. Overall Customer Churn Distribution

```python
churn_counts = df["Churn"].value_counts()

print(churn_counts)
```

# 5. Analyzing Customer Churn Percentage

```python
churn_percentage = df["Churn"].value_counts(normalize=True) * 100

print(churn_percentage.round(2))
```

```python
plt.figure(figsize=(7, 5))

ax = sns.countplot(data=df, x="Churn", width=0.33)

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

# 6. Analyzing Churn by Contract Type

```python
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

print(contract_churn.round(2))
```

```python
# Churn by Contract Type

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Contract', hue='Churn')

plt.title('Customer Churn by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')

plt.tight_layout()
plt.show()
```

# 7. Analyzing Churn by Internet Service

```python
internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

print(internet_churn.round(2))
```

```python
# Churn Rate by Internet Service

internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    data=internet_churn.reset_index(),
    x="InternetService",
    y="Yes", width=0.35
)

plt.title("Churn Rate by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")

# Add values on top of bars
for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", padding=3)

plt.tight_layout()
plt.show()
```

```python
payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

print(payment_churn.round(2))
```

```python
# Churn Rate by Payment Method

payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

plt.figure(figsize=(10, 5))

ax = sns.barplot(
    data=payment_churn.reset_index(),
    x="PaymentMethod",
    y="Yes", width=0.35
)

plt.title("Churn Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Churn Rate (%)")

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", padding=3)

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

# 14. Analyzing Churn by Senior Citizen Status

```python
senior_churn = pd.crosstab(
    df["SeniorCitizen"],
    df["Churn"],
    normalize="index"
) * 100

print(senior_churn.round(2))
```

```python
# Churn Rate by Senior Citizen Status

senior_churn = pd.crosstab(
    df["SeniorCitizen"],
    df["Churn"],
    normalize="index"
) * 100

senior_churn.index = senior_churn.index.map({
    0: "Non-Senior",
    1: "Senior"
})

plt.figure(figsize=(7, 5))

ax = sns.barplot(
    data=senior_churn.reset_index(),
    x="SeniorCitizen",
    y="Yes", width=0.3
)

plt.title("Churn Rate by Senior Citizen Status")
plt.xlabel("Customer Type")
plt.ylabel("Churn Rate (%)")

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", padding=3)

plt.tight_layout()
plt.show()
```

# 17. Comparing Monthly Charges by Churn Status

```python
monthly_charges_churn = df.groupby("Churn")["MonthlyCharges"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)

print(monthly_charges_churn)
```

# 18. Comparing Customer Tenure by Churn Status

```python
tenure_churn = df.groupby("Churn")["tenure"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)

print(tenure_churn)
```

```python
# Churn by Tenure

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x='tenure',
    hue='Churn',
    bins=12,
    multiple='stack'
)

plt.title('Customer Churn by Tenure')
plt.xlabel('Tenure (Months)')
plt.ylabel('Number of Customers')

plt.tight_layout()
plt.show()
```
